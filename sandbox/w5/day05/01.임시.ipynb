```python
PRICE_STANDARD = 0.01   # 쪽당 달러 (Standard)
PRICE_ENHANCED = 0.03   # 쪽당 달러 (Enhanced)
CREDIT = 10.0
USD_KRW = 1400.0

items = [
    ("DOC-HR-014_국내출장_여비_규정_v2.0.pdf", 16),
    ("sample_scanned.pdf", 1),
]

total_pages = 0
for name, pages in items:
	print(f"  {pages:>2}쪽 × ${PRICE_STANDARD:.2f} = ${pages * PRICE_STANDARD:.2f}   {name}")
	total_pages += pages
  
total = total_pages * PRICE_STANDARD
print("  " + "─" * 22)
print(f"  {total_pages:>2}쪽          = ${total:.2f}   합계  ≈ {total * USD_KRW:,.0f}원")
print(f"  ${CREDIT:.0f} 크레딧의 {total / CREDIT * 100:.1f}%")

enhanced = total_pages * PRICE_ENHANCED
print()
print(f"  (Enhanced 로 하면 {total_pages}쪽 × ${PRICE_ENHANCED:.2f} = ${enhanced:.2f}"
      f" ≈ {enhanced * USD_KRW:,.0f}원 · 크레딧의 {enhanced / CREDIT * 100:.1f}%)")
```
