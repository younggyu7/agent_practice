```python
import os                   # 운영체제 관련 표준 라이브러리
import pathlib              # 파일 경로를 객체로 다루는 표준 라이브러리

here = pathlib.Path.cwd()   # 현재 작업 위치 확인

ROOT = here.parents[2] if here.name == "day04" else here

os.chdir(ROOT) # 루트 디렉토리 설정 : 앞으로 모든 상대경로는 이 폴더가 기준이 된다.
print(ROOT)
```

```python
SANDBOX = ROOT / "sandbox" / "w2" / "day04"

# SQLite : 파이썬 표준 라이브러리. 설치 필요없음.
import sqlite3

# 실습용 DB 파일 생성 경로
db_path = SANDBOX / "sql_practice.db"

# db 파일 연결 : 파일이 없으면 새로 만들어짐
conn = sqlite3.connect(db_path)

# 커서 cursor : DB에 명령을 보내고 결과를 받아오는 창구
cur = conn.cursor()

print("연결 완료 : ", db_path)
```
